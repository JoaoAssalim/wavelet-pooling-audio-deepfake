```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
wavelet_results = pd.read_csv("../results/v2/wavelet_results.csv")
average_results = pd.read_csv("../results/v2/average_results.csv")
max_results = pd.read_csv("../results/v2/max_results.csv")
```

```python
max_results["wavelet_family"] = "Max"
average_results["wavelet_family"] = "Average"
```

```python
all_results = pd.concat([wavelet_results, average_results, max_results])
```

```python
all_results = all_results.sort_values(
    ["eer", "roc_auc"],
    ascending=[True, False],
)
```

```python
colors = range(len(all_results))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(
    all_results["wavelet_family"],
    all_results["eer"],
    c=colors,
    cmap="tab20",
)
axes[0].set_title("Evaluation EER by Pooling Method")
axes[0].set_xlabel("Pooling Method / Wavelet Family")
axes[0].set_ylabel("EER (%)")

axes[1].scatter(
    all_results["wavelet_family"],
    all_results["roc_auc"],
    c=colors,
    cmap="tab20",
)
axes[1].set_title("Evaluation ROC AUC by Pooling Method")
axes[1].set_xlabel("Pooling Method / Wavelet Family")
axes[1].set_ylabel("ROC AUC (%)")

for axis in axes:
    axis.tick_params(axis="x", rotation=45)
    axis.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```

```python
colors = range(len(all_results))

plt.scatter(x=all_results["wavelet_family"], y=all_results["eer"], cmap="tab20", c=colors)
plt.title("Wavelet Families Eval EER (%)")
plt.ylabel("EER (%)")
plt.xlabel("Wavelet Families")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
plt.scatter(x=all_results["wavelet_family"], y=all_results["roc_auc"], cmap="tab20", c=colors)
plt.title("Wavelet Families Eval ROC AUC (%)")
plt.ylabel("ROC AUC (%)")
plt.xlabel("Wavelet Families")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
db_16_training = pd.read_csv("../results/v2/training_db16.csv")
db_16_training.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(db_16_training["train_eer"], label="Train EER (%)")
axes[0].plot(db_16_training["dev_eer"], label="Dev EER (%)")
axes[0].set_title("Train and Dev EER - db16")
axes[0].set_ylabel("EER (%)")
axes[0].set_xlabel("Epochs")
axes[0].legend()

axes[1].plot(db_16_training["train_loss"], label="Train Loss")
axes[1].plot(db_16_training["dev_loss"], label="Dev Loss")
axes[1].set_title("Train and Dev Loss - db16")
axes[1].set_ylabel("Loss")
axes[1].set_xlabel("Epochs")
axes[1].legend()

fig.tight_layout()
plt.show()
```
